```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Load model comparison CSV from your local project structure
df = pd.read_csv("../reports/model_comparison.csv")

# Create the bar chart
fig, ax1 = plt.subplots(figsize=(10, 6))
width = 0.35
x = range(len(df["Model"]))

# Plot AUPRC and Flagged Loans (%)
ax1.bar([i - width/2 for i in x], df["AUPRC"], width=width, label="AUPRC")
ax1.bar([i + width/2 for i in x], df["Flagged Loans (%)"], width=width, label="Flagged Loans (%)")

# Labeling
ax1.set_xticks(list(x))
ax1.set_xticklabels(df["Model"], rotation=15)
ax1.set_ylabel("Score / Percentage")
ax1.set_title("Model Performance Comparison")
ax1.legend()

# Save to figures folder
os.makedirs("reports/figures", exist_ok=True)
plt.tight_layout()
plt.savefig("reports/figures/model_performance_comparison.png", dpi=300)
plt.show()
```
